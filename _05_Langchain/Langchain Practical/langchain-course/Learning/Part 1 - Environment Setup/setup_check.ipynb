```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
groq_key = os.environ.get("GROQ_API_KEY")
if groq_key is None:
    raise RuntimeError("GROQ_API_KEY not set -- add it to your .env file (it's free to get).")
print(groq_key[:5])
```

```python
from langchain.agents import create_agent


def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"


# "groq:<model>" is LangChain's provider-shorthand for init_chat_model --
# it reads GROQ_API_KEY from the environment automatically, the same way
# "openai:gpt-4o-mini" would read OPENAI_API_KEY.
agent = create_agent(
    model="groq:openai/gpt-oss-120b",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content_blocks)
```
